```python
# String is Mutable or Immutable?
# Ans: String is Immutable.
# Mutable: insert, delete and update operations can be performed
# Immutable: inset, delete, and update operations cannot be performed.

a="HelloWorld"
a[0]='h'
print(a)
```

## Operations on Strings

### 1. Concatenation (+)

```python
s1="Manohar "
s2="Sai "
s3="Krishna"
print(s1+s2+s3)
```

### 2. Repetition (*)


```python
s="bla "
print(s*6)
```

### 3. Membership (in, not in)


```python
s="Hello World"

print("Hell" in s)
print("hell" in s)
print("o W"in s)
print("o w"in s)
print("Helo" not in s)
print("Hello" not in s)
```

### 4. Indexing & Slicing 

```python
# string[start:end:step]

# default values : start = 0, end = n-1, step = 1

heroine="Samantha"
print(heroine[4])
print(heroine[-1])
print(heroine[-+6])

print(heroine[0:5])
print(heroine[0:5:2])
print(heroine[:5:2])
print(heroine[::2])
print(heroine[::])

print(heroine[6:3:1]) # here step=1, so from start=6 it gets incremented, so there is no chance to go from start=6 to end=3. So it prints empty string.

print(heroine[6:3:-1]) # here step = -1, so from start=6 it gets decremented, means 6->5->4 upto (3-(-1)) = 3 + 1 = 4
print(heroine[-8:])
```

```python
print(heroine[-1:-6]) # here step =1 , so from start=-1, then -1+1 = 0 and then 0+1 = 1, there is no chance to reach -5, so empty string will be printed.
```

```python
print(heroine[-1:-6:-1])
```

```python
print(heroine[-7::])
```

```python
print(heroine[0:-1:])     # It goes from 0 to -1-1=-2 , 0 -> -2

# 0  1  2  3  4  5  6  7
# S  a  m  a  n  t  h  a 
#-8 -7 -6 -5 -4 -3 -2 -1
```

```python
print(heroine[-10:5:1])  # It starts from -8 as -10 is out of range , it goes until 5-1=4, so from -8 -> 4.
# 0  1  2  3  4  5  6  7
# S  a  m  a  n  t  h  a 
#-8 -7 -6 -5 -4 -3 -2 -1
```

```python
# reverse a string
print(heroine[::-1])
```

```python
# Reverse a string without using indexing or shortcut.
# Samantha
rev_str=""  
for char in heroine:
    rev_str=char+rev_str
print(rev_str)
```

```python
rev_str_obj=reversed(heroine)
print(rev_str_obj)
print("".join(rev_str_obj))
```
