## File Handling

```python
import os
os.getcwd()
```

```python
# open('file_name', 'mode')
# f is file object through which we can access the content of the file.
f=open('DS20_file.txt','r')

for line in f:
    print(line)

f.close() # closing the file, deletes from the RAM
```

```python
# with statement: this will automatically close the file

with open('DS20_file.txt','r') as f:
    for line in f:
        print(line)
```

```python
for line in f:
    print(line)
# We cannot access the file here because it has already been closed automatically using 'with'.
```

```python
# read(): to get/read all the lines from the file without using the loop
# read(): Reads the entire content of the file and returns it as a single string.
# It does not require a loop to read all the file content.

with open('DS20_file.txt','r') as f:
    print(f.read())  # reads the entire file
```

```python
# readlines: to get all the lines into a list or to access particular line data
# readlines(): Reads all lines from the file and returns them as a list.
# Each line becomes a separate element in the list.

with open('DS20_file.txt','r') as f:
    lines=f.readlines()
    print(lines)
```

```python
# Accessing a particular line from the list:
# Since list indexing starts from 0, lines[1] represents the second line in the list returned by readlines().

with open('DS20_file.txt','r') as f:
    lines=f.readlines()
    print(lines[1]) 
```

```python
# *lines: Unpacks the list and passes each line as a separate argument to print().
# Therefore, each line is printed separately instead of displaying the list format.

with open('DS20_file.txt','r') as f:
    lines=f.readlines()
    print(*lines) # unpacking the complete list
```

## write() function

```python
# to write the content inside the file
# we will use the 'w' mode with write() function
# if file do not exist then 'w' mode will create a new file.
# read mode 'r' do not create the file if the file do not exist.
```

```python
# 'w' mode opens the file for writing.
# If the file already exists, its previous data will be OVERWRITTEN with the new data given.

with open('pawankalyan123.txt','w') as f:
    f.write("Pawan Kalyan is the brother of Chiranjeevi")
```

```python
with open('pawankalyan123.txt','r') as f:
    print(f.read())
```

```python
# 'w' mode again opens the same file for writing.
# Since the file already exists, the previous data is OVERWRITTEN.

with open('pawankalyan123.txt','w') as f:
    f.write("Pawan Kalyan is the Deputy Chief Minister of Andhra Pradesh State.")
```

```python
with open('pawankalyan123.txt','r') as f:
    print(f.read())
```

## append mode: a

```python
with open('pawankalyan123.txt','a') as f:
    f.write("\nPawan Kalyan is the Leader of JanaSena Party")
```

```python
with open('pawankalyan123.txt','r') as f:
    print(f.read())
```

## writelines: writing multiple lines at the same time in the file

```python
with open('pawankalyan123.txt','a') as f:
    f.writelines(["\n His nickname is Power Star", "\n Born on September 2, 1971, in Andhra Pradesh"])
# as we are using the append 'a' mode, it will append to the previous data without overwriting it.
```

```python
with open('pawankalyan123.txt','r') as f:
    print(f.read())
```

```python
# w+ mode: write/read mode
# w+ mode will clear the previous content
# w+ create a new file if file doesn't exist
```

```python
with open('pawankalyan123.txt','w+') as f:
    f.write("Pawan Kalyan as OG")
    print(f.read())
```

## Why doesn't read() print anything after write() in w+ mode?

```python
with open('pawankalyan123.txt','w+') as f:
    f.write("Pawan Kalyan as OG")
    f.seek(0)
    print(f.read())
```

```python
# r+: read/write mode
# file must exist, otherwise shows error

with open('pawankalyan123.txt','r+') as f:
    f.write("Pawan Kalyan is an hard working person.")
    f.seek(0)
    print(f.read())
```

```python
with open('pawankalyan456.txt','r+') as f:   # here the file 'pawankalyan456.txt' does not exist.
    f.write("Pawan Kalyan is an hard working person.")
    f.seek(0)
    print(f.read())
```
