# 2. Bitwise Operators ( <<, >>, ~, &, |, ^)


```python
# Bitwise Operators works on binary numbers
# Bitwise Operators are faster than arithmetic operators
# Bitwise Operators perform many operations in easy way due to binary number system
# Proof: 50*64 => 50+50+50+50+50+50+50+50+........................+50 (Arithmetic Operation)
#            Total= 63 operations
#    50<<6=> 50 << 6 = 50 * 2^6  = 50*64 = 3,200. (Bitwise Left shift Operation)
```

```python
a=10
b=3
c=2
print(a&b)
# 8 4 2 1
# 10= 1010
# 2 = 0010
# 3 = 0011
# 10 & 3=0010 = 2
print(a|b)
# 10 | 3 = 1011 = 11
print(a^b)
# 10 ^ 3 = 1001 = 9
print(a^b^c)
# If we have odd no of 1's then the result is 1, if we have odd no of 1's then the result is 0.

print(~b) # ~x = -(x+1)


```

# Left Shift(<<), Right Shift(>>)

```python
# Left Shift : Inserts new bits(0) to the Least Significant Bit(LSB)
# Right Shift : Deletes the bits from the Least Significant Bit(LSB)

# a << k = a*(2^k)
```

```python
# Left Shift(<<) ( Number is doubled for every left shift)
# a << k = a*(2^k)

a=10                     # 128 64 32 16 8 4 2 1
print(a<<1) # 10 = 0000 1010, 10<<1 = 0001 0100 = 20 (or) 10*(2^1)=10*2=20
```

```python
print(a<<2) # 10<<2 = 0000 1010 becomes 0010 1000 = 40 (or) 10*(2^2) = 10*4 = 40
```

```python
print(a<<6) # 10<<6 = 0000 0000 1010 becomes 0010 1000 0000 = 2⁹ + 2⁷ = 512 + 128 = 640 (or) 10*(2^6) = 10*64 = 640
```

```python
print(a<<8) # 10<<8 = 0000 0000 1010 becomes 1010 0000 0000 = 2¹¹ + 2⁹ = 2048 + 512 = 2560 (or) 10*(2^8) = 10*256 = 2650
```

```python
# Right Shift(>>) (Number is halved for every right shift)
# a >> k = floor( a/(2^k) ) 

a=64
print(a>>1)

print(10>>1) # 1010 >> 1 = 0101 = 5
```

```python
print(a>>2)

print(10>>2) # 1010 >> 2 = 0010 = 2
```

```python
print(a>>3)

print(10>>3) # 1010 >> 3 = 0001 = 1 
print(10>>4)
print(10>>5)
```

```python
b=10
print(b>>2) # 10>>2 = floor[10/(2^2)] = floor[10/4] = floor[2.5] = 2 
```

# Bitwise NOT Operator (~)

```python
#def: Flip the bits
# ~x = -(x+1)

a=5
print(~a)
```

```python
# Negative Number
a=-11
print(~a) # -(-11+1) = -(-10) = 10
```
