# Functions

```python
# def: A function is a block of reusable code that perform a specific task.

# defining the function
def func():
    print("Welcome to functions")

func() # calling the function
func()
```

```python
# Passing parameter to the function

def func(num):
    print(f'Value: {num}')

func(10)
func(20)
```

```python
def func(num1, num2):
    print(f'Value 1: {num1}')
    print(f'Value 2: {num2}')
    
func(10,20)
```

```python
# Example: nested functions (function inside another function)

def outer():
    print("I am outer function")
    def inner():
        print("I am inner function")
    inner()
outer()
```

```python
# return statement
def func(a,b):
    return a+b
    
sum=func(1,2)
print(sum)

print(func(10,20))
```

```python
# return multiple values from functions using 'return'
def func(a,b):
    return a+b, a-b
    
sum, diff = func(20, 5)

print(sum)
print(diff)
```

```python
# not returning a value in function, but still accessing the value from function call will give 'None'.

def get_age():
    return

print(get_age())
```

```python
def ds_func():
    lst1=[10,20,30,40,]
    tuple1=(50,60,70,80)
    return lst1, tuple1
print(ds_func())
```

```python
# returning a function from other function

def outer():
    def inner():
        print("I am inner function")
    print(inner)   # This will print the address of inner function ( Purpose of this line: To understand how a function returns another function.)
    # Note: Printing function name will print the address where it stored.
    return inner   # This will return the inner function.  


f = outer() # here f stores the reference address of the inner function because outer() function is returning the inner() function.
print(f)    # This is same as the line: print(inner), in the outer() function
f()         # This will call the inner() function because f stores the inner() function address.
```

```python
# Default Parameters
# Defining a default value for the parameter.

def func(a='Nothing'):
    print(a)
    
func(10)
func()
```

```python
def addition(a=0, b=0):
    print(a+b)
    
addition(10,20)
addition(10)
addition()
```

```python
def addition(a=10, b=20):
    print(a+b)
    
addition(30,40)

addition(30)
addition(a=30) # both are same because addition(30) will pass 30 to the first variable 'a=30', and the default value for b is 20, a+b = 30+20 =50.

addition(b=30)

addition()
```

## Passing arbitrary number of arguments using *args

```python
# Use *args when we do not know in advance how many parameters we will pass to our function.

def func(*args):  # *args , we can write anything in place of args (eg: *values)
    for i in args:
        print(i, end=" ")
    print()

func() # prints nothing
func("Manohar")
func("Manohar","Sai")
func("Manohar","Sai","Krishna")
```

```python
# In Python **kwargs allows a function to accept any number of keyword arguments (named inputs with a key=value structure).
# It stands for "keyword arguments".
# The double asterisk (**) tells Python to pack all extra named inputs into a regular dictionary, 
# where the argument names become the dictionary keys and the values become the dictionary values.

def build_profile(**kwargs):
    print(kwargs)  # Inside the function, 'kwargs' is a dictionary
    for key, value in kwargs.items():
        print(f"{key}: {value}")

build_profile(name="Alice", role="Engineer", location="Remote")
```
