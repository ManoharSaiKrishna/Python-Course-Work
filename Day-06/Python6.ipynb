```python
Biryani = input("Enter the Famous Hyderabad Biryani") 

if Biryani=="Paradise":
    print("Spicy & Tasty")
elif Biryani=="Shahdab":
    print("Double Masala")
elif Biryani=="Road Side":
    print("Budget Friendly")
else:
    print("PG Biryani")
```

```python
# Replacing the elif with else: if

Biryani = input("Enter the Famous Hyderabad Biryani") 

if Biryani=="Paradise":
    print("Spicy & Tasty")
else:
    if Biryani=="Shahdab":
        print("Double Masala")
    else:
        if Biryani=="Road Side":
            print("Budget Friendly")
        else:
            print("PG Biryani")
```

```python
# Ternary condition for the above code. We have to write the printing statement first and then the condition.
print("Spicy & Tasty") if Biryani=="Paradise" else print("Double Masala") if Biryani=="Shahdab" else print("Budget Friendly") if Biryani=="Road Side" else print("PG Biryani") 
```

```python
# Structured representation of above ternary condition
Biryani = input("Enter the Famous Hyderabad Biryani") 
(print("Spicy & Tasty") if Biryani=="Paradise" else 
 print("Double Masala") if Biryani=="Shahdab" else 
 print("Budget Friendly") if Biryani=="Road Side" else 
 print("PG Biryani") )
```

# match case


```python
# Match case is used to replace multiple if else conditions
# Match case is most powerful than the switch case statement
# Match case works with variables and darta structures
```

```python
# Example-1
num=int(input("Enter a number"))

match num:
    case 10:
        print("You entered 10")
    case 20:
        print("You entered 20")
    case 30:
        print("You entered 30")
    case _:
        print("Something else")
```

```python
Biryani = input("Enter the Famous Hyderabad Biryani") 

match Biryani:
    case "Paradise":
        print("Spicy & Tasty")
    case "Shahdab":
        print("Double Masala")
    case "Road Side":
        print("Budget Friendly")
    case _:
        print("PG Biryani")
```

```python
# Example-2 : Passing data structure to match case

data=[1,2,3,4]

match data:
    case [4,3,2,1]:
        print("Not Matched")
    case [1,2,3,4]:
        print("Matched")
```

```python
# Example-3: if condition with match case

val = int(input("Enter a number"))

match val:
    case x if x>0:
        print("It is a positive number")
    case y if y<0:
        print("It is a negative number")
    case _:
        print("It is Zero")
```

```python
val = int(input("Enter a number"))

match val:
    case x if x>0:
        print("It is a positive number")
    case x if x<0:
        print("It is a negative number")
    case _:
        print("It is Zero")
```

```python
data=[1,2,3,4]

match data:
    case x if 2 in data:
        print("List contains 2")
    case x if 2 not in data:
        print("List do not contain 2")
```
