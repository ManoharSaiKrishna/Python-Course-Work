# List Data Structure

## Delete items from the list (remove(), pop(), clear(), del )

```python
food_list=["Pappu", "Biryani", "Fried Rice", "Sambar", "Payasam", "Curd Rice"]

# remove(): deletes the item that is mentioned from the list. It remove first match only.
food_list.remove("Fried Rice")
print(food_list)

# pop() is used to delete the item from a particular index. If the index is not mentioned then the last item will be deleted.
food_list.pop()
print(food_list)

food_list.pop(2)
print(food_list)

# pop() can also returns the removed item
removed_item=food_list.pop(1)
print(removed_item)

```

```python
# del is capable of deleting a slice range of multiple items at once.
# IndexError if the specified index is out of bounds

letters=['a','b','c','d','e','f']
del letters[1]
print(letters)
del letters[1:4]
print(letters)

del letters[2:4] # But this will not raise an error, because we are using slicing here
print(letters)

del letters[:]   # This will clear the list which is same as letters.clear()
print(letters)  


# del letters[4]
# print(letters)  This will raise IndexError as the index is Out of Bounds

# Python slicing and slice deletion never raise an IndexError when indices go out of bounds. 
# While deleting a single index (like del letters[10]) will crash your program,

# del letters   // This will deletes the list itself. so print(letters) will raise an error that ` name 'letters' is not defined `
```

```python
# clear() is used to delete all the items from the list

letters.clear()
print(letters)
```

## Popular List inbuilt functions

```python
### a=[22,23,22,19,18,21,16,19,20,99]
print(len(a))
print(min(a))
print(max(a))
print(sum(a))

a.reverse()
print(f'a: {a}')

# list.sort() modifies the original list in-place and returns None, 
# while the sorted() function returns a new sorted list, leaving the original data completely unchanged.
# Both Python's list.sort() method and the built-in sorted() function use an algorithm called Timsort.
# Timsort is a hybrid, stable sorting algorithm. It is derived from a combination of Merge Sort and Insertion Sort.
# The Best Case TC=O(n), Average Case & Worst Case TC = O(nlogn)

b=sorted(a)
print(f'a: {a}')
print(f'b: {b}')
# we can also use reverse=True with sorted() function
c=sorted(a, reverse=True)
print(f'c: {c}')
a.sort()
print(f'a: {a}')

a.sort(reverse=True)
print(f'a: {a}')
```

```python
# list() function

t1=(10,20,30,40,50)
s1={60,80,70,30}

l1=list(t1)
l2=list(s1)

value= range(1,10)
l3=list(value)

print(l1)
print(l2)
print(l3)
```

## List comprehension

```python
# creating a list in a single line using a loop

lst = [i for i in range(1,11)]
print(lst)

lst1 = [i**2 for i in range(1,11)]
print(lst1)

lst3 = [i for i in range(1,11) if i%2==0]
print(lst3)
# This is equivalent to the below code.

lst4=[]
for i in range(1,11):
    if(i%2==0):
        lst4.append(i)
print(lst4)
```

```python
l1=[i for i in range(1,11) if i%2==0 if i%3==0]
print(l1)

# This code is equivalent to the below code.

l2=[]
for i in range(1,11):
    if i%2==0:
        if i%3==0:
            l2.append(i)
print(l2)
```

```python
l3=[i**2 if i%2==0 else i+2 for i in range(1,11)]
print(l3)
# 1-> 1+2=3, 2-> 2^2=4, 3-> 3+2=5, 4-> 4^2=16, 5-> 5+2=7, 6-> 6^2=36, 7-> 7+2=9, 8-> 8^2=64, .....

# This code is equivalent to the below code.

lst4=[]
for i in range(1,11):
    if i%2==0:
        lst4.append(i**2)
    else:
        lst4.append(i+2)
print(lst4)
```

```python
l1=[i**2 if i%2==0 else i+2 for i in range(1,11) if i>5 and i<9]
print(l1)

# This code is equivalent to the below code.

l2=[]
for i in range(1,11):
    if i>5 and i<9:
        if i%2==0:
            l2.append(i**2)
        else:
            l2.append(i+2)
print(l2)

# i=6,7,8 values will be satisfied by i>5 and i<9.
# so 6%2==0-> 6^2=36, 7%2!=0-> 7+2=9, 8%2==0-> 8^2=64.
```
