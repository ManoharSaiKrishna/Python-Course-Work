# Loops

## for loop

```python
animals=['Lion','Tiger','Cheetah','Dinosaur','Dog']
for i in animals:
    print(i)
```

```python
food_name="Chicken_Dum_Masala_Biryani"
lst=food_name.split('_')
for i in food_name:
    print(i)
for i in lst:
    print(i)
```

```python
# for loop with range() 
# range() function is used to generate list of elements
# syntax of range: range(start, end, step) => it will go from start to end-1
# by default the value of start in range() function is 0 and step is 1
# in for loop the incrementation will be done automatically where as in while loop we have to explicitly increment.
for i in range(10):
    print(i)
```

```python
for i in range(4,10):
    print(i)
```

```python
for i in range(2,10,2):
    print(i)
```

```python
# break statement
# break statement will bring the control out of the loop
for num in range(1,6):
    if num==3:
        break
    print(num)
```

```python
# continue statement
# continue statement takes the control to the start of the loop

for num in range(1,6):
    if num==3:
        continue
    print(num)
```

```python
# nested loops
for i in range(1,4):
    for j in range(1,3):
        print(i,j)
```

```python
for i in range(3,6):
    for j in range(2,5):
        print(i,j)
```

```python
# else statement
# else statement will execute if the loop is executed entirely without executing the break statement in it.

for i in range(5):
    if i==2:
        continue
    elif i==4:
        break
    else:
        print(i)
else:
    print("Loop executed successfully")
```

```python
for i in range(5):
    if i==2:
        continue
    elif i==40:
        break
    else:
        print(i)
else:
    print("Loop executed successfully")
```

```python
# pattern
n=int(input("Enter a number: "))
for i in range(n):
    print('* '*n)
```

```python
n=int(input("Enter a number: "))
for i in range(n):
    for j in range(n):
        print('*', end=" ")
    print()
```

```python
n=int(input("Enter a number: "))
for i in range(1,n+1):
    print("* "*i)
```

```python
n=int(input("Enter a number: "))
for i in range(1,n+1):
    for j in range(i):
        print('*', end=" ")
    print()
```

```python
n=int(input("Enter a number: "))
for i in range(n,0,-1):
    for j in range(i):
        print('*', end=" ")
    print()
```

```python
n=int(input("Enter a number: ")) # Method 1
for i in range(1,n+1):
    left_spaces=" "*(n-i)
    stars="* "*i
    print(left_spaces+stars)
    
```

```python
n=int(input("Enter a number: ")) # Method 2
for i in range(1,n+1):
    left_spaces=""
    stars=""
    for j in range(n-i):
        left_spaces+=" "
    for j in range(i):
        stars+="* "
    print(left_spaces+stars)
    
```

```python
n=int(input("Enter a number: ")) # Method 3
for i in range(n,0,-1):
    left_spaces=" "*(i-1)
    stars="* "*(n-i+1)
    print(left_spaces+stars)
```
