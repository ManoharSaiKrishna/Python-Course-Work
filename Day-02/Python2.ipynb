# Python Indentation

```python
x=10
if x==10:
    print(x)
print("Hello")
```

```python
age=20
if age >= 18:
  print("Adult")
  print("Can vote")
# Both statements use 2 spaces, so they are in the same block.
# We can use single, double, triple spaces but it should be consistent across the block
```

```python
if age >= 18:
  print("Adult")
  if age >= 21:
    print("Can enter")
# Here: 2 spaces → first if block
# 4 spaces → nested if block
# Python doesn't require exactly 4 spaces; 2 spaces can work too, as long as the indentation is consistent.
```

# Type Casting

```python
# Integer TypeCasting: int()
a = int(5.6) # float to int
b=int('300') # string to int
c=int(True) # bool to int
d=int(False)
print(a,type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
print(int("101",2)) # Binary string to integer
print(int("101",10)) # Decimal string to integer
```

```python
# Float TypeCasting: float()

a=float(105)
b=float("105")
c=float("10.2")
d=float(True)
e=float(False)

print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
print(e, type(e))
```

```python
# Convert to string: str()

a=str(200)
b=str(4.5)
c=str(True)

print(a, type(a))
print(b, type(b))
print(c, type(c))
```

```python
# Convert to Boolean: bool()

a=bool(1)
b=bool(0)
c=bool(12)
d=bool(a)
e=bool("Hello")


print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
print(e, type(e))
```
