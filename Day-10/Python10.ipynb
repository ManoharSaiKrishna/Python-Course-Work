## Updating List Items

```python
# updating single element

lst=["Prabhas", "Mahesh", "Ram Charan", "Pawan Kalyan", "NTR", "Allu Arjun"]
lst[1]="Dance with"
print(lst)
```

```python
# updating multiple elements: update more elements with less no of elements

lst[3:5]=["Dance with"]
print(lst)
```

```python
# updating less elements with more elements

lst[1:4]=["Yash", "Karthi", "Vijay", "Suriya"]
print(lst)
```

```python
lst[-2]="Dance With"
print(lst)
```

```python
lst[-5:-1]=["Dance with"]
print(lst)
```

# inserting items in list

```python
# append(), extend(), insert()

# append() function is used to insert a single element at the end of the list.

l1=[]
l1.append(10)
print(l1)
l1.append(20)
print(l1)
l1.append(30)
print(l1)
l1.append("Fourty")
print(l1)
```

```python
# extend() function is used to insert multiple elements at the end of the list.
# the extend() method adds all the individual elements of an iterable (such as a list, tuple, set, or string) to the end of an existing list.

l1.extend([50,60,70]) # list of elements are added to the end of existing list.
print(l1)
l1.extend((80,90))    # tuple of elements are added to the end of existing list.
print(l1)
```

```python
# insert(index, value) function is used to insert at specific index

l1.insert(3,40)
print(l1)
```

```python
l1.insert(3,[32,34,36])
print(l1)
```
