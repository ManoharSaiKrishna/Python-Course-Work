# OOPS: Object Oriented Programming System

```python
# Example-1: structure of class

# class is a keyword, food is the class name
# class variable: the value of class variable remains same for all the objects.
# class variable value cannot be changed by object.
# class variable value can only be changed by class name.


class food:
    food_name="Biryani"  # class variable

obj=food()   # creating the object(obj) of food() class
print(obj.food_name) # accessing the class variable

obj2=food()
print(obj2.food_name)

# accessing the class variable by using the class name

print(food.food_name)

# changing the class variable using the class name
food.food_name="Mutton"
print(food.food_name)

obj3=food()
print(obj3.food_name)

# Trying to change the class variable value using the object

obj4=food()
obj4.food_name="Pulao" # Here obj4 is creating its own local variable with the same name(food_name)
print(obj4.food_name)
print(food.food_name)  # Here the class variable value is not changed.
```

```python
# Example-2: Constructor function

# __init__() is a constructor method that runs automatically when a new object is created. 
# It is used to initialize object data.
# constructor function is used to initialize the instance variables.

# 'self' refers to the current object, allowing each object to store and access its own data.

# class variable value is same for all the objects.
# instance variable values are different for every object

# class Bike: creates a class named Bike, which acts as a blueprint for bike objects.
class Bike:
    no_tyres=2      # class attribute / class variable

    def __init__(self,company, model, price):
        print("This is Constructor Function")
        self.company_name=company  # Instance Variable / Instance Attribute
        self.model=model      # Instance Variable / Instance Attribute
        self.price=price
        print("Price: ",self.price) # Printing the instance variable in the constructor function.

# self.company and self.model are instance attributes, unique to each Bike object created from the class.
        
obj1=Bike("Yamaha", "RX100", 70000)
print(obj1.company_name," : ", obj1.model)
print()
obj2=Bike("Bajaj", "NS160",150000)
print(obj2.company_name," : ", obj2.model)

print("No.of tyres:",obj2.no_tyres)
```

```python
# Example-3: Member function

# Member function do not get called automatically, we need to call them explicitly by ourself.
# We need to call them explicitly using the object.


class Bike:
    no_tyres=2      

    def __init__(self,company, model, price):
        self.company_name=company  
        self.model=model      
        self.price=price

    # Member function
    def display(self):
        print("The price of the bike: ",self.price)
 
obj1=Bike("Yamaha", "RX100", 70000)
print(obj1.company_name," : ", obj1.model)

# display() is a member function.
# It is NOT called automatically when obj1 is created.
# We explicitly call it using obj1.display()
obj1.display() 

print()

obj2=Bike("Bajaj", "NS160",150000)
print(obj2.company_name," : ", obj2.model)

# Explicitly calling the member function for obj2
obj2.display()


```

```python
# code for the above question

class SmartWallet:
    def __init__(self, balance):
        self.balance = balance
        self.count = 0

    def process(self, amount):
        fee = 5 if (self.count + 1) % 3 == 0 else 0
        if self.balance + amount - fee >= 0:
            self.balance += (amount - fee)
            self.count += 1

balance =int(input())
txs = list(map(float, input().split()))
wallet = SmartWallet(balance)
for t in txs:
    wallet.process(t)
print(f"{wallet.balance:.2f}")
```

## Encapsulation

```python
# Def: Encapsulation refers to the binding of data(attributes) and methods(functions) into a single unit (class),
# We can access the variables/data/attributes through object outside the class
# We can access the variables/data/attributes through functions defined inside the class.
# Encapsulation helps organize related data and behavior together and provides a way to control how the data is accessed and modified.

class Bike:
    no_tyres=2      # class variable / attribute / data

    def __init__(self,company, model, price):
        self.company_name=company  
        self.model=model      
        self.price=price

    # Member function
    def display(self):
        print("The price of the bike: ",self.price)
 
obj1=Bike("Yamaha", "RX100", 70000) 
print(obj1.company_name," : ", obj1.model)
obj1.display()
print(obj1.no_tyres)

# Here:
# Data/attributes: no_tyres, company_name, model, price
# Methods: __init__() and display()
# Both data and methods are contained/bound within the Bike class → this is encapsulation.
```

## Access Specifiers / type of attributes in encapsulation

```python
class Marvel:
    def __init__(self,name,power,weapon):
        self.Name=name     # Public instance variable
        self._Power=power  # Protected instance variable
        self.__Weapon=weapon # Private instance variable

    def display(self):
        print(self.Name, self._Power, self.__Weapon) # Accessing the Public, Protected and Private variables inside the class.

obj1=Marvel("Thor",1000,"Hammer")
obj1.display()

print(obj1.Name)     # Accessing the Public variable outside the class.
print(obj1._Power)   # Accessing the Protected variable outside the class.
print(obj1._Marvel__Weapon) # Accessing the Private variable outside the class.

# Syntax for accessing the private variable: obj._classname__privateVariable

# Protected variables can be accessed outside the class, but we should not access.
# In python public, protected, private concept is used just for Sign/Warning.
# But they are fully implemented in java, restricting the users to access the protected and private variable outside the class.
```

## Inheritance

# Single Inheritance: One parent to One child


```python
# parent
class Restaurant:
    taste='very very very spicy'
    def Biryani(self):   # Here, self represents the object that calls the method.
        print('Very very tasty Biryani')

# child class
class PistaHouse(Restaurant):
    pass

child_obj=PistaHouse()  # It is an object of child class

# child class object can access the properties and menthods of the parent class.

print(child_obj.taste)
child_obj.Biryani()
Restaurant.Biryani(child_obj) # These two are same.

# Restaurant.Biryani() - This gives error because Biryani() is an instance method, and you are calling it using the class name without providing an object.

# When you do this: child_obj.Biryani()
# Python internally treats it approximately as: Restaurant.Biryani(child_obj)
# So self = child_obj, and it works.
# Because PistaHouse inherits from Restaurant, child_obj can access the inherited method.
```

```python
# super() function
# super(): It is used to access the properties & methods from a parent (superclass) inside a child (subclass).

# parent class
class Restaurant:
    taste='very very very spicy'
    def Biryani(self):   
        print('Very very tasty Biryani')

# child class
class PistaHouse(Restaurant):
    def display(self):
        super().Biryani()    # accessing the parent class method in the child class using super() keyword. 
        print(super().taste) # accessing the parent class variable in the child class using super() keyword.

child_obj = PistaHouse()  # It is an object of child class

child_obj.display()
```

```python
# parent class
class Restaurant:
    taste='very very very spicy'
    def __init__(self):        # constructor function of parent class 
        print("Restarurant Class Constructor Function")
    def Biryani(self):    
        print('Very very tasty Biryani')

# child class
class PistaHouse(Restaurant):
    def __init__(self):      # constructor function of child class
        print("PistaHouse Class Constructor Function") 
    
child_obj = PistaHouse()  # It is an object of child class

# if a child class have constructor function, when we create an object of child class then the constructor function of child class will get called automatically, 
# the constructor function of parent class won't get called further.
```

# Multiple Inheritance: Many Parents to One child


```python
class A:
    def __init__(self):
        print("Class A Constructor")
    def display(self):
        print("A")

class B:
    def __init__(self):
        print("Class B Constructor")
    def show(self):
        print("B")

class C(A,B):
    def execute(self):
        super().display()
        super().show()
     
obj_C = C() # This will call the constructor function of class A when the object is created.
obj_C.execute()

```

```python
# Method Resulution Order(MRO) -> Example-1

class A:
    def __init__(self):
        print("Class A Constructor")
    def display(self):
        print("A")

class B:
    def __init__(self):
        print("Class B Constructor")
    def display(self):
        print("B")

class C(A,B):
    def display(self):
        super().display()
       
obj_C = C()
obj_C.display()
print(f'Method resolution Order of C Class: {C.mro()}')

```

```python

# Method Resulution Order(MRO) -> Example-2

class A:
    def __init__(self):
        print("Class A Constructor")
    def display(self):
        print("A")

class B:
    def __init__(self):
        print("Class B Constructor")
    def display(self):
        print("B")

class C(A,B):
    def display(self):
        super().display()

class D(B,A):
    def display(self):
        super().display()
        

obj_C = C()
obj_C.display()
print(f'Method resolution Order of C Class: {C.mro()}')

print()

obj_D = D()  # # This will call the constructor function of class B when the object is created (This is according to MRO).
obj_D.display()
print(f'Method resolution Order of D Class: {D.mro()}')

```

```python
# Example : super()
# super() follows the MRO Approach
# MRO: Method Resolution Order

class A:
    def display(self):
        print("I am A Class Constructor")
        super().display() # This will call the display() function of Class B (According to MRO of C)

class B:
    def display(self):
        print("I am B Class Constructor")

class C(A,B):
    def hello(self):
        super().display()  # This will call the display() function of Class A (According to MRO of C)

obj_C = C()
obj_C.hello()


# MRO : Method Resolution Order

print([i.__name__ for i in C.__mro__])
```

# Multilevel Inheritance

```python
# Great GF -> Grand Father -> Father -> Son

class GreatGF:
    def gold(self):
        print("1000 kg")

class GrandFather(GreatGF):
    def land(self):
        print("500 Acres")

class Father(GrandFather):
    def car(self):
        print("Range Rover")

class Son(Father):
    def bike(self):
        print("Java Bike")

son_obj = Son()
son_obj.bike()
son_obj.car()
son_obj.land()
son_obj.gold()
print()
father_obj = Father()
father_obj.car()
father_obj.land()
father_obj.gold()
# father_obj.bike() -> father cannot access the methods of son class
    
```

```python
# Great GF -> Grand Father -> Father -> Son

class GreatGF:
    def __init__(self):
        print("I am GreatGF Constructor")
    def gold(self):
        print("1000 kg")

class GrandFather(GreatGF):
    def __init__(self):
        print("I am GrandFather Constructor")
    def land(self):
        print("500 Acres")

class Father(GrandFather):
    # def __init__(self):
    #     print("I am Father Constructor")
    def car(self):
        print("Range Rover")

class Son(Father):
    # def __init__(self):
    #     print("I am Son Constructor")
    def bike(self):
        print("Java Bike")

son_obj = Son()
son_obj.bike()
son_obj.car()
son_obj.land()
son_obj.gold()
print()
father_obj = Father()
father_obj.car()
father_obj.land()
father_obj.gold()
    
```

```python
# Example: Method Overriding

class GreatGF:
    def car(self):
        print("Ambassador")

class GrandFather(GreatGF):
    def car(self):
        print("Maruti 800")

class Father(GrandFather):
    def car(self):
        print("Tata Car")

class Son(Father):
    '''
    def car(self):
        print("Lambhorgini")    
    '''

son_obj = Son()
son_obj.car()
    
```

```python
# super() function

class GreatGF:
    def gold(self):
        print("1000 kg")

class GrandFather(GreatGF):
    def land(self):
        print("500 Acres")
        super().gold()

class Father(GrandFather):
    def car(self):
        print("Range Rover")
        super().land()

class Son(Father):
    def bike(self):
        print("Java Bike")
        super().car()

son_obj = Son()
son_obj.bike()
```

# Hierarchial Inheritance - One Parent Many Child

```python
# Base Parent Class
class Animal:
    def eat(self):
        print("Animal is eating.")
    def sleep(self):
        print("Animal is sleeping.")

# Derived Child Class 1
class Dog(Animal):
    def bark(self):
        print("Dog says: Bow Bow!")

# Derived Child Class 2
class Cat(Animal):
    def meow(self):
        print("Cat says: Meow!")


dog = Dog()
cat = Cat()

dog.eat()    
dog.bark()   

cat.sleep()  
cat.meow()   
```

```python
# Example : MRO

class Restaurant:
    def __init__(self):
        print("Restaurant Constructor")

class Mehfil(Restaurant):
    def __init__(self):
        print("Mehfil Constructor")

    def mehfil_menu(self):
        print("Mehfil serves Biryani and Kebabs")


class PistaHouse(Restaurant):
    '''
    def __init__(self):
        print("PistaHouse Constructor")
        '''

    def pista_menu(self):
        print("Pista House serves Haleem and Biryani")


# Creating objects
obj1 = Mehfil()
obj1.mehfil_menu()
print([i.__name__ for i in Mehfil.__mro__])
print()

obj2 = PistaHouse()
obj2.pista_menu()
print([i.__name__ for i in PistaHouse.__mro__])
```

```python
# Different function/method names in each class to understand super().
class A:
  A_var="A class"
  def A_output(self):
    print("I am A class")

class B(A):
  B_var="B class"
  def B_output(self):
    print("I am B class")

class C(B):
  C_var="C class"
  def C_output(self):
    print("I am C class")

class D(B):
  D_var="D class"
  def D_output(self):
    print("I am D class")
    
class E(C,D):
  E_var="E class"
  def E_output(self):
    print("I am E class")
    super().C_output()
    super().D_output()
    super().B_output()
    super().A_output()

obj_1=E()


obj_1.E_output()


print([i.__name__ for i in E.__mro__])
```

```python
# Same function/method names in each class to understand super() and MRO.

class A:
  A_var="A class"
  def A_output(self):
    print("I am A class")

class B(A):
  B_var="B class"
  def A_output(self):
    print("I am B class")
    super().A_output()

class C(B):
  C_var="C class"
  def A_output(self):
    print("I am C class")
    super().A_output()

class D(B):
  D_var="D class"
  def A_output(self):
    print("I am D class")
    super().A_output()
      
    
  
class E(C,D):
  E_var="E class"
  def A_output(self):
    print("I am E class")
    super().A_output()
      
obj_1=E()
obj_1.A_output()

print([i.__name__ for i in E.__mro__])
print([i.__name__ for i in C.__mro__])
```
