## Creating different types of tuples

```python
# empty tuple
empty_tuple=()
e2=tuple()

print(empty_tuple)
print(type(empty_tuple))
print()
print(e2)
print(type(empty_tuple))
```

```python
a=(1,2,3,(4,5,6))
print(a, type(a))

print(a[3], type(a[3]))
```

```python
# we can also store a list, set, tuple and dict in a tuple

a=(1,[2,3],{4,5},(6,7),"Eight",{'c':9})
print(a, type(a))
print(a[1], type(a[1]))
print(a[2], type(a[2]))
print(a[3], type(a[3]))
print(a[5], type(a[5]))
```

```python
# converting list, set into a tuple

lst=[1,2,3,4]
s={1,2,3,4}
a=tuple(lst)
b=tuple(lst)
print(a)
print(b)

# converting dictionary into a tuple.

dict1={'b':5, 'c':6}
print(tuple(dict1)) 
# When you convert a dictionary directly into a sequence (like a tuple or a list), 
# Python only iterates over the keys by default. It ignores the values unless you explicitly ask for them.
print(tuple(dict1.keys()))
print(tuple(dict1.values()))
print(tuple(dict1.items()))
```

## Accessing the tuple items

```python
a=(1,2,3,4,5,6,7,8)
print(a[3])
print(a[-2])
print(a[1:5])
print(a[::])
print(a[::-1]) # reverse a tuple
```

# Insert, delete and update operations indirectly on tuple 

```python
t1=(10,20.5,"Hello", [10,20,30])
print(t1[3][2])
```

```python
tuple[0]=1000
```

```python
t1[3][2]=1000
print(t1)

# We can change the items of a list inside a tuple directly because a list is a mutable (changeable) object, 
# even when it is nested inside an immutable (unchangeable) tuple.
# While we cannot swap out the list for a completely different list, you can freely modify, add, or remove elements 
# within that existing list using standard list operations.
```

```python
# This will fail and throw a TypeError:
t1[3] = [4, 5, 6] 
```

```python
# The other way to update tuple elements is to convert it into a list, do the operations and then again convert the list into a tuple.

# insert, remove, delete

tuple1=(1,2,3)

list1=list(t1)  # converting a tuple into a list
print(list1) 

list1.append(5) # and then doing operations on list items.
print(list1)

list1.remove(3)
print(list1)

del list1[0]
print(list1)

list1.pop()
print(list1)

list1.insert(0,1)
print(list1)

list1.extend([3,4,5])
print(list1)

tuple1=tuple(list1) # again converting the list into a tuple.
print(tuple1)

```

## Popular Built-in functions for tuple

```python
a=(6,70,100,600,700,90,10,80,120)

print(len(a))
print(min(a))
print(max(a))
print(sum(a))

sorted_tuple1=sorted(a) # here sorted() funtions returns a list of elements. so we need to explicitly convert it into a tuple.
# a.sort()  // The tuple object does not have a .sort() method because tuples are immutable
# The .sort() method sorts a list 'in place', which changes the original data.
#  Because tuples cannot change size or content, Python blocks any methods that try to modify them.

print(sorted_tuple1)
print(tuple(sorted_tuple1))

# print(sorted_tuple2)
```

```python
# reversing a tuple
a=(6,70,100,600,700,90,10,80,120)

print(reversed(a)) 
print(tuple(reversed(a)))

# sorting a tuple in descending order.

a=sorted(a,reverse=True) # this returns a list.
a=tuple(a)
print(a)
```
