```python
# capturing (or accessing) the exception object

try:
    print(y)
except Exception as e:
    print(f"The error is: {e}")
    print(type(e))
    
print("Hello")
a,b=10,20
print(a+b)
```

```python
# else will be executed if there is no error in the code written in the try block.
try:
    print(y)
except Exception as e:
    print(f"The error is: {e}")
    print(type(e))
else:
    print("Yay!, We successfully executed the code without having an error")
    
print("Hello")
a,b=10,20
print(a+b)
```

```python

try:
    y=20
    print(y)
except Exception as e:
    print(f"The error is: {e}")
    print(type(e))
else:
    print("Yay!, We successfully executed the code without having an error")
    
print("Hello")
a,b=10,20
print(a+b)
```

```python
# finally block will be executed regardless of having an error or not in the try block.


try:
    y=20
    print(y)      # It will be executed without an error.
except Exception as e:
    print(f"The error is: {e}")
    print(type(e))
else:
    print("Yay!, We successfully executed the code without having an error")
finally:
    print("I will be execute whatever happens")
    
print("Hello")
a,b=10,20
print(a+b)
```

```python
# finally block will be executed regardless of having an error or not in the try block.


try:
    print(z)   # It will raise an error, because 'z' is not defined
except Exception as e:
    print(f"The error is: {e}")
    print(type(e))
else:
    print("Yay!, We successfully executed the code without having an error")
finally:
    print("I will be execute whatever happens")
    
print("Hello")
a,b=10,20
print(a+b)
```

## Types of errors

```python
# NameError: Occurs when you use a variable or function that doesn't exist.
try:
    print(age)
except NameError as e:
    print(f"Error: {e}")
```

```python
# TypeError: Occurs when you perform an operation using incompatible data types.

try:
    result = 10 + "20"
except TypeError as e:
    print(f"Error: {e}")
```

```python
# ValueError: Occurs when the type is correct but the value is inappropriate.

try:
    age = int("hello")
except ValueError as e:
    print(f"Error: {e}")

# int() can accept a string, but "hello" cannot be converted into an integer.
# Remember: ValueError → Right type, wrong value
```

```python
# IndexError: Occurs when you try to access a list index that doesn't exist.

try:
    numbers = [10, 20, 30]
    print(numbers[5])
except IndexError as e:
    print(f"Error: {e}")
```

```python
# KeyError: Occurs when you try to access a dictionary key that doesn't exist.

try:
    student = {"name": "Manohar"}
    print(student["age"])
except KeyError as e:
    print(f"Error: {e}")
```

```python
# ZeroDivisionError: Occurs when we try to divide by zero.

try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"Error: {e}")
```

```python
# AttributeError: Occurs when you try to access an attribute or method that an object doesn't have.

try:
    name = "Python"
    name.append("!")
except AttributeError as e:
    print(f"Error: {e}")
```

```python
# Syntax Error: Occurs when we break Python's grammar/rules. 
try:
    if 10 > 5
        print("Hello")
except SyntaxError as e:
    print(f"Error: {e}")
```

## use of raise keyword

```python
# 'raise' is used to generate user defined errors
# The raise keyword in Python is used to manually trigger an exception or error.

age = -20
if age<0:
    raise Exception("Sorry, Age cannot be negative")
```

```python
def set_age(age):
    if age < 0:
        # Manually trigger a ValueError
        raise ValueError("Age cannot be negative!")
    print(f"Age set to {age}")

try:
    set_age(-5)
except ValueError as error:
    print(f"Caught an error: {error}")
```
