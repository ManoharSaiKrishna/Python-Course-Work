## Creating different types of sets

```python
empty_set=set()

set1=set([10,20,30,])
set2=set((10,20,30))
set3={1,2,3,4}
print(set1)
print(set2)
print(set3)
```

```python
# frozenset is immutable (cannot perform inser, selete, update operations)

fro_set=frozenset([10,20,30,40])
print(fro_set)
```

```python
# we can only store immutable items in a set.
# we cannot store mutable items in a set.
# so we can store tuple, frozenset in a set as they are immutable.
# we cannot store list, set, dictionary in a set, because they are mutable.

set1={[10,20,30], "Hello", 40}  # storing a list in a set.
print(set1)
```

```python
set2={(10,20,30), "Hello", 40} # storing tuple in a set
print(set2)
```

```python
set3={{10,20,30}, "Hello", 40} # storing a set in a set.
print(set3)
```

```python
set4={frozenset([10,20,30]), "Hello", 40} # storing fronzenset in a set.
print(set4)
```

## Insertion in set ( add(), update())

```python
# The add() method inserts one item into the set. If the item is already present, the set remains unchanged.

set1={10,20.5,"Hello"}

set1.add("world")
print(set1)
```

```python
set1.add([10,20,30]) # we cannot add a list, dictionary or a set directly into a set. It will raise a TypeError because they are mutable and unhashable.
```

```python
set1.add((10,20,30)) # here we are adding a entire tuple into the set. Since tuple is immutable this will not raise an error.
print(set1)
```

```python
# To insert multiple items at once, use the update() method. 
# It accepts any iterable object, such as lists, tuples, dictionaries, or other sets.

fruits = {"apple", "banana"}

# Add items from a list
fruits.update(["mango", "grape"])
print(fruits)  
# Output: {'banana', 'mango', 'apple', 'grape'} // order may differ since set is unordered

# Add items from another set
fruits.update({"pineapple", "papaya"})
print(fruits)

# Add items from a tuple
fruits.update((10,20)) # this will add individual items into the set.
print(fruits)

```

```python
# Alternative: Set Union Operator (|=)
# We can also use the in-place union operator (|=) to merge another set into your current set. 
# It behaves identically to update().

a = {10, 20}
a |= {30, 40}
print(a)
# Output: {10, 20, 30, 40}
```

## Delete operation on set (remove(), discard(), pop(), clear())

```python
a={1,2,3,4,5,6,7,8,9}
```

```python
# Use remove() if you are certain the item exists.
# remove() function deletes an item which is mentioned.
# if item is missing it will raise 'KeyError'.
# it will not return any value
a.remove(9)
print(a)
# a.remove(10)  // This will raise error because 10 is not present in the set
```

```python
# discard() don't raise an error if item is not present in the set.
# Use discard() to delete safely without errors.
a.discard(10)
print(a)
```

```python
# pop() - removes a random element from the set because set is unordered.
print(a)
a.pop()
print(a)
```

```python
# clear() -> clears the set.

a.clear()
print(a)
```

```python
del a  # deletes the set
# print(a)
```

## Popular Built-in functions in sets

```python
set1={10,20,40,2,3,5,60,70,1000}

print(len(set1))
print(min(set1))
print(max(set1))
print(sum(set1))
```

```python
l1=sorted(set1) # It returns a list.
print(l1)
print(set(l1)) # If we convert the list into a set again after sorting, the items will not be in sorted order because set is unordered.
# set do not have .sort() function
```

## Membership operation on sets

```python
# membership operators are used to test whether a specific element exists within a collection. 
# When used with sets, Python evaluates these operations significantly faster than with lists or tuples 
# because sets utilize hash tables, allowing for O(1) constant time complexity lookups on average.

fruits = {"apple", "banana", "cherry"}

print("apple" in fruits)   # Output: True

print("orange" in fruits)  # Output: False

print("orange" not in fruits)  # Output: True

print("banana" not in fruits)  # Output: False

```
