```python
# Receiving Multiple Inputs
a,b,c = input("Enter 3 names").split() # ram sam ajay => ['ram', 'sam', 'ajay'] => a,b,c = 'ram', 'sam', 'ajay'
print(a,b,c)
```

```python
a,b,c = input("Enter 3 names").split(',') # in this case Input: ram,_sam,_ ajay, then after splitting with reference to ',' extra space will be added to each name 
print(a,b,c)
```

```python
a,b,c = input("Enter 3 names").split(',') # In this case we are splitting the string with reference to the ',' but in the input we are giving the names with space seperated, so this will raise an error
print(a,b,c)
```

```python
x=list(map(int, input("Enter 3 numbers").split(','))) # 1,2,3 => split(',') => ['1','2','3'] => map(int, ['1','2','3']) => [1, 2, 3]
print(x)
x1, x2, x3 = x
print(x1,x2,x3)
```

# Output in Python

```python
#1. Concatenation using commas
#2. String concatenation using '+'
#3. str.format() method
#4. f-string method (Python 3.6+ version)
#5. Escape Sequences
#6. End and Sep parameters in print()
```

```python
#1. Concatenation using commas

hero_name='Prabhas'
age=46
fee='200cr'
print(hero_name, age, fee)
```

```python
#2. String concatenation using '+'

hero_name='Prabhas'
heroine_name='Anushka'
fee=200

print(hero_name + " " + heroine_name + " " + str(fee))
print(hero_name + " " + heroine_name + " " + fee)
```

```python
#3. str.format() method

hero_name= 'Ram Charan'
age=40
print("Name: {}\nAge : {}".format(hero_name, age))
print("Name: {}, Age:{}".format(hero_name, age))
print("Name: {0}, Age:{1}".format(hero_name, age))
print("Name: {1}, Age:{0}".format(hero_name, age))
```

```python
#4. f-string method (Python 3.6+ version)

hero_name= "Prabhas"
age=40

print(f"Name of the hero: {hero_name}, Age: {age}")
```

```python
hero_name= "Prabhas"
heroine_name="Anushka"
fee=100

# print(f"Name of the hero: "{hero_name}" \nHeroine Name: {heroine_name} \nFee: {fee}") => this will raise an error because it understand that the string is ended ( f"Name of the hero: "), that's why we use escape character '\' to use inner " inside the f-string

print(f"Name of the hero: \"{hero_name}\" \nHeroine Name: {heroine_name} \nFee: {fee}")
print(f'Name of the hero: "{hero_name}", Heroine Name: {heroine_name}, Fee: {fee}') # here we are using f'' to be able to use "" inside it
```

```python
#6. End and Sep parameters in print()

# end parameter is used to decide how the print statement will END.
# sep parameter decides how the multiple values will be seperated in a line.

hero_name= "Prabhas"
heroine_name="Anushka"
fee=100

print(hero_name)
print(heroine_name)

print(hero_name, end="\n") # This is same as "print(hero_name)" because the default value for end is '\n'.
print(heroine_name)

print(hero_name, end=" ")
print(heroine_name)

print(hero_name, end=", ")
print(heroine_name)
```

```python
print(hero_name, heroine_name, fee)
print(hero_name, heroine_name,fee, sep=" ") # Both of these are same, because the default value for sep is " ".
print(hero_name, heroine_name,fee, sep=",") 
```
