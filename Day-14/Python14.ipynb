# Dictionary Data Structure

## Different ways to create dictionary:

```python
empty_dict={}
empty_dict2=dict()

print(empty_dict)
print(type(empty_dict))

print(empty_dict2)
print(type(empty_dict2))
```

```python
int_dict={1:"Mango", 2:"Orange", 3:"Watermelon"} # integers, floats, strings, tuples are allowed as keys in a dictionary, because they are immutable.
print(int_dict)

float_dict={1.2:"Mango", 2.5:"Orange", 3.6:"Watermelon"}
print(float_dict)

str_dict={'A':"Mango", 'B':"Orange", 'C':"Watermelon"}
print(str_dict)

hetero_dict={1:"Mango", 2.5:"Orange", 'C':"Watermelon"}
print(hetero_dict)

tuple_dict={(1,2):"Mango", (3,4):"Orange", 'C':"Watermelon"}
print(tuple_dict)

# list_set_dict={[1,2]:"Mango", {3,4}:"Orange", 'C':"Watermelon"}    // list and sets as a key is not allowed in dictionary, because lists and sets are mutable,
```

```python
# Dictionary Comprehension

dict_comp={x:x**2 for x in range(1,6)}
print(dict_comp)
```

```python
# Duplicate Keys - example

int_dict={1:"Mango", 1:"Orange", 3:"Watermelon"}
print(int_dict)
```

```python
list_of_tuples=[('Manohar',6),('Sai',90),('Krishna',100)]

dict1=dict(list_of_tuples)
print(dict1)
```

```python
list_of_lists=[['Manohar',6],['Sai',90],['Krishna',100]]

dict1=dict(list_of_lists)
print(dict1)
```

```python
tuple_of_tuples=(('Manohar',6),('Sai',90),('Krishna',100))

dict1=dict(tuple_of_tuples)
print(dict1)
```

```python
tuple_of_lists=(['Manohar',6],['Sai',90],['Krishna',100])

dict1=dict(tuple_of_lists)
print(dict1)
```

```python
set_of_lists={['Manohar',6],['Sai',90],['Krishna',100]} # Lists cannot be stored in a set, because list is a mutable object.

dict1=dict(set_of_lists)
print(dict1)
```

```python
set_of_tuples={('Manohar',6),('Sai',90),('Krishna',100)} 

dict1=dict(set_of_tuples)
print(dict1)
```

```python
# Accessing dictionary items

drink_items={"Thumbsup":50, "Sugarcane":20, "CoconutWater":60, "Water":20, "Rosemilk":70}
print(drink_items["Water"])
```

```python
print(drink_items["Artos"]) # dict[key] will raise an error if it is not present in the dictionary.
```

```python
print(drink_items.get("Artos")) # dict.get(key) will return None if the mentioned key is not present in the dictionary.
```

```python
print(drink_items.get("Artos", "Mentioned Drink is not avilable"))
```

```python
print(drink_items.items())   # returns the items in the dictionary in the format of list of tuples [(),(),(),..].
for item in drink_items.items(): 
    print(item)
```

```python
for i in drink_items: # returns the keys in the dictionary
    print(i)
```

```python
print(drink_items.keys()) # returns the keys in the dictionary as a list.
for i in drink_items.keys(): 
    print(i)
```

```python
print(drink_items.values()) # returns the values in the dictionary as a list.
for i in drink_items.values(): 
    print(i)
```

```python
# nested dictionary

nested_dict={
        1: {'Hello':10, 'World':20},
        2: {'A': 50, 'B':60}
       }
print(nested_dict)
print(nested_dict[1])
print(nested_dict[1]['World'])
```

## Insert & Update operations on dictionary

```python
x_dict={'name':'Manohar','age':20}

x_dict['age']=21  # update
print(x_dict)

x_dict['city']="RJY"
print(x_dict)
```

```python
# insert and update using the update() function
# update() can insert and update multiple items together at a time

x_dict.update({'age':25, 'occupation':'IAS', 'name':'Sai', 'Salary':6}) # here we are updating the value of the keys age, name and simultaneously we are inserting new key-value pairs occupation & salary.
print(x_dict)
              
```

## Delete the items from the dictionary

```python
# pop(), popitem(), clear(), del()

x_dict
```

```python
# pop() remove the item whose key is mentioned

x_dict.pop('Salary')
print(x_dict)
```

```python
# popitem() will delete the last element and returns the removed item

removed_item = x_dict.popitem()
print(x_dict)
print(removed_item)
```

```python
# clear() function clears the dictionary

x_dict.clear()
print(x_dict)
```

```python
# del will delete the entire dictionary

del x_dict
# print(x_dict) // This will raise the error because the dictionary is already deleted.
```
