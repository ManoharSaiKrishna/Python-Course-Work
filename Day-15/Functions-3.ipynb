## enumerate function

````python
fruits_names=['orange','guava','banana'] # we can use enumerate with lists, tuples, strings.
for fruit in fruits_names:
    print(fruit)
print()
for index, fruit in enumerate(fruits_names):
    print(f'{index} : {fruit}')
print()
for index, fruit in enumerate(fruits_names, start=101):
    print(f'{index} : {fruit}')
````

## zip() function:

````python
# Example: Usage of @ to represent the decorator function

def decorator_func(func):     
    def wrapper_func():         
        print("Buy the required ingredients") 
        func()
        print("Add Cream")   
        print("Add Cherries")
    return wrapper_func

@decorator_func
def cake():
    print("Take the flour")
    print("Add eggs and the other ingredients")
    print("Put the cake in the oven")

cake()
````

````python

### 1. First, this function is created

```python
def cake():
    print("Take the flour")
    print("Add eggs")
    print("Put the cake in the oven")
```

At this point:

```text
cake → original cake function
```

---

### 2. Then Python sees

```python
@decorator_func
def cake():
    ...
```

The `@` means:

```python
cake = decorator_func(cake)
```

Notice: **there are no `()` after `wrapper_func` here.**

Python calls:

```python
decorator_func(cake)
```

Inside `decorator_func`:

```python
def decorator_func(func):
```

So:

```text
func → original cake function
```

Then it creates:

```python
def wrapper_func():
    print("Buy the required ingredients")
    func()
    print("Add Cream")
    print("Add Cherries")
```

And returns the **function itself**:

```python
return wrapper_func
```

Therefore:

```python
cake = decorator_func(cake)
```

becomes effectively:

```python
cake = wrapper_func
```

So now:

```text
cake ─────────→ wrapper_func
                  │
                  ↓
             func → original cakes
```

---

## 3. Now `cake()` is called

You write:

```python
cake()
```

But `cake` now points to `wrapper_func`.

So this is equivalent to:

```python
wrapper_func()
```

The wrapper executes:

```python
print("Buy the required ingredients")
```

Then:

```python
func()
```

Since `func` points to the **original cake**, the original cake executes.

Then:

```python
print("Add Cream")
print("Add Cherries")
```

---

## Important: `cake = wrapper_func()` vs `cake = wrapper_func`

You specifically asked:

> how `cake = wrapper_func()`

Actually, **it is NOT `cake = wrapper_func()`**.

It is:

```python
cake = wrapper_func
```

### Difference:

```python
cake = wrapper_func
```

means:

> Make `cake` refer to the `wrapper_func` function.

Whereas:

```python
cake = wrapper_func()
```

means:

> Execute `wrapper_func` immediately and store its return value in `cake`.

In your example, we want the **function itself**, so the decorator does:

```python
return wrapper_func
```

not:

```python
return wrapper_func()
```

### One-line mental model

```text
@decorator_func
cake
   ↓
cake = decorator_func(cake)
   ↓
cake = wrapper_func
   ↓
cake()
   ↓
wrapper_func() executes
   ↓
func() → original cake() executes
```

**The decorator doesn't execute `cake` when decorating it; it replaces `cake` with `wrapper_func`.**
````

````python
# Example: Usage of @ to represent the decorator function

def decorator_func1(func):     
    def wrapper_func():         
        print("Buy the required ingredients") 
        func()
    return wrapper_func

def decorator_func2(func):     
    def wrapper_func():         
        func()
        print("Add Cream")   
        print("Add Cherries")
    return wrapper_func

@decorator_func1
@decorator_func2

def cake():
    print("Take the flour")
    print("Add eggs and the other ingredients")
    print("Put the cake in the oven")

cake()
````
