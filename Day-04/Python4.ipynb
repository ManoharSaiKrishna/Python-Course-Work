 # 3. Assignment Operators (+=, -=, *=, /=, &=, |=, <<=, >>=, ^=, :=, ....)


```python
a=10

a+=5   # a = a + 5
print(a)

a-=5   # a = a - 5
print(a)

a*=5   # a = a * 5
print(a)

a/=5   # a = a / 5
print(a)
print(int(a))

a=int(a)
print(a)

a%=2   # a = a % 2
print(a)

```

```python
a=10
a&=3 # a = a & 3
print(a)
```

```python
# Walrus Operator  (:=)
# Walrus Operator will first assign the value to the variable then pass it to outer function

# print(x=2) this will raise the error

print(x:=2) # Where as this won't raise an error because walrus operator is executed first by assigning the value to x and then prints the value of x
```

# 4. Comparison Operators ( ==, >=, <=, !=, <, >)

```python
a=10
b=5

print(a==b)
print(a>=b)
print(a<=b)
print(a!=b)
print(a<b)
```

# 6. Logical Operators ( and, or, not)

```python
a=True
b=False
print(a and b)
print(a or b)
print(not a)
print(not b)
```

# 7. Identify Operators (is, is not) 

```python
a=[1,2,3]
b=[1,2,3]
c=a

print(a==b)
print(a is b)
print(a is c)
print(a is not c)
```

# Membership Operators (in, not in)

```python
fruits=['orange', 'banana', 'sapota', 'watermelon']
print('orange' in fruits)
print('banana' not in fruits)
```
