# Loops

## while loop

```python
# Example-1

i=1
while(i<=10):
    print(i)
    i+=1
print(f"i: {i}")
```

```python
# Example-2 : break statement
# break statement helps us to come out of the loop

i=1
while i<=10:
    print(i)
    if i==5:
        break
    i+=1
    
```

```python
# continue 
# Difference between pass and continue: 
# The core difference is that pass does absolutely nothing and allows code execution to continue sequentially, while 
# continue stops the current loop iteration and immediately jumps back to the top of the loop for the next cycle
```

```python
i=1
while i<10:
    i+=1
    if i<5:
        continue
    print(i)
```

```python
i=1
while i<10:
    i+=1
    print(i)
    if i<5:
        continue
    
```

```python
i=1
while i<10:
    if i<5:
        continue
    i+=1
    print(i)  # this will enter into infinite loop because when i=1 ,the condition i<5 will be true which will use continue statement. So this makes the loop to jump back to the top of the loop. This is infinite loop. 

# Because this code creates an infinite loop, executing it will freeze the notebook. You will need to restart the Jupyter kernel to stop it.
  
```

```python
# Example-4: else statement with while loop
# else statement exeecutes only if loop is completely executed successfully without executing any break statements in the loop.

i=1
while i<=10:
    print(i)
    i+=1
    if i>5:
        break
else:
    print("Loop is successfully completed")
```

```python
i=1
while i<=10:
    print(i)
    i+=1
else:
    print("Loop is successfully completed")
```

```python
for i in range(5):
    print(i)
else:
    print("Loop completed")
# Here, the loop finishes normally so else statement executes.
```

```python
for i in range(5):
    print(i)
    if i == 3:
        break
else:
    print("Loop completed")
# Here, break stops the loop → else does not execute.
```

```python
# Example-5: pass statement
# pass statement is a placeholder for future code
# pass is a do-nothing statement. It is used when Python requires a statement, but you don't want to execute any action yet.

i=0
while i<=10:
    if i==5:
        # This will raise the error , so we will use pass statement in this condition
    else:
        print(i)
    i+=1
```

```python
i=0
while i<=10:
    if i==5:
        pass
    else:
        print(i)
    i+=1
```
