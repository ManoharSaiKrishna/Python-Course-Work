# Function Overloading

```python
# This type of function overloading is supported in C++ & Java but not in Python
# Based on no. of arguements we are passing to the function, C++ & Java can differentiate the functions. 

def func(var1):
    print(var1)
def func(var1,var2):
    print(var1, var2)
func(10)
```

## Using Default Arguements for function overloading

```python
# here var2 is the default parameter/argument whose default value is 'None'

def func(var1, var2=None):
    if var2 is not None:
        print(var1, var2)
    else:
        print(var1)
        
func(10)
func(10,20)
```

```python
def func(var1, var2=None, var3=None):
    if var3 is not None:
        print(var1, var2,var3)
    elif var2 is not None:
        print(var1,var2)
    else:
        print(var1)
        
func(10)
func(10,20)
func(10,20,30)
```

```python
# Example: Using the default arguments to implement Function Overloading

def calculator(a=0,b=0,c=0,d=0):
    return a+b+c+d

print(calculator())
print(calculator(10))
print(calculator(10,20))
print(calculator(10,20,30))
print(calculator(10,20,30,40))
```

## Using *args to implement the function overloading

```python

def calculator(*args):
    return sum(args)

print(calculator())
print(calculator(10))
print(calculator(10,20))
print(calculator(10,20,30))
print(calculator(10,20,30,40))
```

## Using Decorator to implement function overloading

```python
from multipledispatch import dispatch

@dispatch(int)
def func(var):
    print(var)

@dispatch(str)
def func(var):
    print(var)

@dispatch(int,int)
def func(var1, var2):
    print(var1, var2)

@dispatch(float,float)
def func(var1, var2):
    print(var1, var2)

@dispatch(float,float,float)
def func(var1, var2, var3):
    print(var1, var2, var3)

func(10)
func('Manohar')
func(10,20)
func(11.1,20.3)
func(11.1, 20.3, 51.2)
```

# Operator Overloading

```python
# we can perform different operations on the user defined objects using operator overloading.
```

```python
class Dosa:
    def __init__(self,variety,price):
        self.Variety=variety
        self.Price=price

d1=Dosa("Masala",60)
d2=Dosa("Onion",50)

print(d1+d2)

# This is not valid
```

```python
# Arithemetic Operators
class Dosa:
    def __init__(self,variety,price):
        self.Variety=variety
        self.Price=price
    def __add__(self,d2):
        total_price=self.Price+d2.Price
        return total_price
    def __sub__(self,d2):
        total_price=self.Price-d2.Price
        return total_price
    def __mul__(self,d2):
        total_price=self.Price*d2.Price
        return total_price
    def __truediv__(self,d2):
        total_price=self.Price/d2.Price
        return total_price
    def __floordiv__(self,d2):
        total_price=self.Price//d2.Price
        return total_price
    def __mod__(self,d2):
        total_price=self.Price%d2.Price
        return total_price
    def __pow__(self,d2):
        total_price=self.Price**d2.Price
        return total_price


d1=Dosa("Masala",25)
d2=Dosa("Onion",10)

print(d1+d2)  # __add__ (These all are the built-in method names to overload the respective operator)
print(d1-d2)  # __sub__
print(d1*d2)  # __mul__
print(d1/d2)  # __truediv__
print(d1//d2) # __floordiv__
print(d1%d2)  # __mod__
print(d1**d2) # __pow__
```

```python
# Comparison Operators Overloading

class Dosa:
    def __init__(self,variety,price):
        self.Variety=variety
        self.Price=price
    def __gt__(self,d2):
        total_price=self.Price>d2.Price
        return total_price
    def __lt__(self,d2):
        total_price=self.Price<d2.Price
        return total_price
    def __eq__(self,d2):
        total_price=self.Price==d2.Price
        return total_price
    def __ge__(self,d2):
        total_price=self.Price>=d2.Price
        return total_price
    def __le__(self,d2):
        total_price=self.Price<=d2.Price
        return total_price


d1=Dosa("Masala",25)
d2=Dosa("Onion",10)

print(d1>d2)  # __gt__
print(d1<d2)  # __lt__
print(d1==d2) # __eq__
print(d1>=d2) # __ge__
print(d1<=d2) # __le__
```

## Data Abstraction

```python
# Example-1 : how to define abstract class, abstract method.

from abc import ABC, abstractmethod

# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): # abstract method
        pass

class Hyd_Mahindra(Head_Mahindra): # child class
    pass

obj_hyd=Hyd_Mahindra()  # instantiating/creating an object of Hyd_Mahindra child class
```

```python
# Example-1 : how to define abstract class, abstract method.

from abc import ABC, abstractmethod

# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): # abstract method
        pass

class Hyd_Mahindra(Head_Mahindra): # child class of abstract class
    def air_bag(self): # abstract method
        print("I will use nylon to make the Air bag")

class Noida_Mahindra(Head_Mahindra): # child class
    pass

obj_hyd=Hyd_Mahindra()  # instantiating/creating an object of Hyd_Mahindra child class
obj_noida=Noida_Mahindra()
```

```python

from abc import ABC, abstractmethod

# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): # abstract method
        pass      

class Hyd_Mahindra(Head_Mahindra): # child class of abstract class
    def air_bag(self): # abstract method
        print("I will use nylon to make the Air bag")

class Noida_Mahindra(Head_Mahindra): # child class
    def air_bag(self): # abstract method
        print("I will use Cotton to make the Air bag")

obj_hyd=Hyd_Mahindra()  # instantiating/creating an object of Hyd_Mahindra child class
obj_noida=Noida_Mahindra()
```

```python

from abc import ABC, abstractmethod

# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): # abstract method
        pass      
        
    @abstractmethod
    def LED_Screen(self):
        pass

class Hyd_Mahindra(Head_Mahindra): # child class of abstract class
    def air_bag(self): # abstract method
        print("I will use nylon to make the Air bag")
     
    def LED_Screen(self):
        print("Sony LED Screen")

class Noida_Mahindra(Head_Mahindra): # child class
    def air_bag(self): # abstract method
        print("I will use Cotton to make the Air bag")
    def LED_Screen(self):
        print("Samsung LED Screen")

obj_hyd=Hyd_Mahindra()  # instantiating/creating an object of Hyd_Mahindra child class
obj_noida=Noida_Mahindra() # instantiating/creating an object of Noida_Mahindra child class

```

```python

from abc import ABC, abstractmethod

# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): # abstract method
        pass      
        
    @abstractmethod
    def LED_Screen(self):
        pass

    def insurance(self): # concrete method
        print("You can take Mahindra company insurance")
        
class Hyd_Mahindra(Head_Mahindra): # child class of abstract class
    def air_bag(self): # abstract method
        print("I will use nylon to make the Air bag")
     
    def LED_Screen(self):
        print("Sony LED Screen")

class Noida_Mahindra(Head_Mahindra): # child class
    def air_bag(self): # abstract method
        print("I will use Cotton to make the Air bag")
    def LED_Screen(self):
        print("Samsung LED Screen")

obj_hyd=Hyd_Mahindra()  # instantiating/creating an object of Hyd_Mahindra child class
obj_noida=Noida_Mahindra() # instantiating/creating an object of Noida_Mahindra child class

obj_hyd.insurance()
```

```python
# A concrete method in an abstract class can be overridden by its child class, just like a normal inherited method.

class Head_Mahindra(ABC):

    def insurance(self):       # Concrete method
        print("Mahindra company insurance")

class Hyd_Mahindra(Head_Mahindra):

    def insurance(self):       # Overriding concrete method
        print("Hyderabad unit insurance")
obj_hyd = Hyd_Mahindra()
obj_hyd.insurance()
```
