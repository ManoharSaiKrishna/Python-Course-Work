# Lambda Function

```python
# Normal Function
def square(x):
    return x * x

print(square(5))
```

```python
# Same thing using lambda

square = lambda x: x * x
print(square(5))
```

```python
func = lambda x: x + 10
print(func(20))

# means: Take x as an argument and return x + 10.
```

```python
# Two arguments:
add = lambda a, b: a + b

print(add(10, 20))
```

```python
# Check whether a number is even:

is_even = lambda x: x % 2 == 0

print(is_even(8))
```

```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))

print(squares)
```

```python
# Two arguments:
lambda a, b: print(a + b)

add(10, 20)
```

```python
# Three arguments:

prod = lambda a,b,c: a*b*c
print(prod(2,3,4))
```

```python
# using lambda function inside the other function

def func(p):
    obj = lambda a:a**p
    return obj

obj1 = func(2) # p = 2
print(obj1(3)) # a = 3, a:a**p => 3**2 = 9
print(obj1(4))
```

```python

def func(p):
    obj = lambda a:a**p if a%2==0 else a+p
    return obj

obj1 = func(2) 

print(obj1(2))  # 2**2 = 4
print(obj1(3))  # 3 + 2 = 5
print(obj1(4))  # 4**2 = 16
```

```python
# sort the given list of tuples w.r.t marks
# each tuple has (name, marks)

students = [("Manohar",80), ("Sai",75), ("Krishna",90)]

students.sort() # This will sort the tuples in the list based on the names. Since 'name' is in index=0, the tuples will be sorted based on the 'name' by default.
print(students)
```

```python
# sort the given list of tuples w.r.t marks
# using the lambda function inside the sorted function

students = [("Manohar",80), ("Sai",75), ("Krishna",90)]

sorted_data = sorted(students, key = lambda a:a[1])
print(sorted_data)
```

```python
students = [("Manohar",80), ("Sai",75), ("Krishna",90)]

sorted_data = sorted(students, key = lambda a:a[1], reverse=True)
print(sorted_data)
```

```python
# sort the given list of tuples w.r.t marks
# using the lambda function inside the .sort() function

students = [("Manohar",80), ("Sai",75), ("Krishna",90)]

students.sort(key = lambda a:a[1])
print(students)
```

```python
# Each tuple has (name, marks, age)
# Now sort the tuples in the list based on the age.

students = [("Manohar",80, 21), ("Sai",75,20), ("Krishna",90, 23)]

sorted_data = sorted(students, key = lambda a:a[2]) # here age is in 2nd index, so we index=2 as a key for sorting the tuples.
print(sorted_data)
```

```python
students = {"Manohar":80, "Sai":75, "Krishna":90}
data = students.items() # .items() will return the key-values pairs of a dictionary as a list of tuples 
print(data)
sorted_data = sorted(data, key = lambda a:a[1])
print(sorted_data)
```

## Recursive Functions

```python
# Example-1

def foo(i):
    if i<1:       # Base Condition / Termination Condition
        return
    print(i)
    foo(i-1)
foo(6)
```

```python
# Example-2

def foo(i):
    if i<1:
        return
    foo(i-1)
    print(i)
foo(6)
```

```python
# Example-3

def foo(i):
    if i<1:
        return
    print(i)
    foo(i-1)
    print(i)
foo(6)
```

```python
# Example-4

def foo(i):
    if i<1:
        return
    foo(i-1)
    print(i)
    foo(i-1)
foo(3)
```
